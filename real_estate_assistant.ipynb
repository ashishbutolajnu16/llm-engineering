# This is a real estate assistant which will give top 5 best properties on the basis of budget

```python
import os
import json
from dotenv import load_dotenv
from IPython.display import Markdown, display, update_display
from scraper import fetch_website_links, fetch_website_contents
from openai import OpenAI
```

```python
load_dotenv(override=True)
api_key = os.getenv('OPENAI_API_KEY')

if not api_key:
    print("No API key was found")
elif not api_key.startswith("sk-proj-"):
    print("API key doesn't start with sk-proj-")
else:
    print("API key looks good")
```

```python
MODEL = 'gpt-5-nano'
openai = OpenAI()
```

```python
real_estate_system_prompt = """
You are an real estate assistant in Dehradun. You will analyze differente website on internet
on the basis of provided location in Dehradun(E.g. Ranipokhri, Doiwala) budget(30 Lakhs, 20 Lakhs),
area(100 Gaj, 200 Gaj etc.),. You need to show top 5 property to the customer.
Include details info of the property with address and maybe plot image. Why should you buy and maybe some pros and cons and maybe provide the 
seller number so customer can contact him
"""
```

```python
customer_prompt = """
Here is the requirement of the customer.
Location: Ranippokhri,
Area: 100 Gaj
Budget: 20-30 Lakhs.
Use above info to get the top 5 properties in this area
"""
```

```python
def stream_brochure():
    stream = openai.chat.completions.create(
        model="gpt-4.1-mini",
        messages=[
            {"role": "system", "content": real_estate_system_prompt},
            {"role": "user", "content": customer_prompt}
        ],
        stream=True
    )
    response = ""
    display_handle = display(Markdown(""), display_id=True)
    for chunk in stream:
        response += chunk.choices[0].delta.content or ''
        update_display(Markdown(response), display_id=display_handle.display_id)
```

```python
stream_brochure()
```
